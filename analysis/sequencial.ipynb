```python
from matplotlib import pyplot as plt
import numpy as np

from pathlib import Path
```

```python
path = next(c for c in (Path("output.csv"), Path("../analysis/output.csv"), Path("analysis/output.csv")) if c.exists())

text = path.read_text()
numbers = np.array(text.replace("(", " ").replace(")", " ").replace(",", " ").split(), dtype=float)
viewport = numbers.reshape(720, 1280, 3)

plt.imshow(viewport, origin="lower", )
plt.show()
```
